```python
class Student:
    def __init__(self,name,roll_no,branch):
        self.name = name
        self.roll_no = roll_no
        self.branch = branch
    def greet(self):
        print(f"hello {self.name} your roll no is {self.roll_no} and your branch is {self.branch}")
        print("hello my name is :-",self.name)
        print("my roll no is :-",self.roll_no)
        print("my branch is :-",self.branch)
        


obj =Student(name = "anshika" , roll_no = 101 , branch = "cse")
obj.greet()
```

```python
class Dog:
    def __init__(self,name):
        self.name = name
    def bark(self):
        print(f"woof woof {self.name}")




dog1=Dog(name = "tommy")

dog2=Dog(name = "Buddy")
```

```python
dog1.bark()
```

```python
dog2.bark()
```
