```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs

# Generate multiple datasets (blobs of points)
X1, _ = make_blobs(n_samples=100, centers=[(5, 5)], cluster_std=0.5, random_state=42)
X2, _ = make_blobs(n_samples=150, centers=[(12, 12)], cluster_std=1.2, random_state=42)
X3, _ = make_blobs(n_samples=80, centers=[(20, 5)], cluster_std=0.8, random_state=42)
X_noise = np.random.uniform(low=0, high=25, size=(30, 2))  # Random noise points

# Combine all datasets
X = np.vstack((X1, X2, X3, X_noise))

# Apply DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)
labels = dbscan.fit_predict(X)

# Plot clusters
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', edgecolors='k', s=50)
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("DBSCAN Clustering on Multiple Datasets")

# Highlight noise points (labeled as -1)
noise_points = X[labels == -1]
if len(noise_points) > 0:
    plt.scatter(noise_points[:, 0], noise_points[:, 1], c='red', marker='x', label='Noise')

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons, make_blobs

# Generate crescent (moon-shaped) data
X1, _ = make_moons(n_samples=200, noise=0.05, random_state=42)

# Generate some circular blobs
X2, _ = make_blobs(n_samples=150, centers=[(-1, 1)], cluster_std=0.3, random_state=42)
X3, _ = make_blobs(n_samples=100, centers=[(2, -0.5)], cluster_std=0.3, random_state=42)

# Add some random noise points
X_noise = np.random.uniform(low=-2, high=3, size=(50, 2))

# Combine all datasets
X = np.vstack((X1, X2, X3, X_noise))

# Apply DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels = dbscan.fit_predict(X)

# Plot clusters
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', edgecolors='k', s=50)
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("DBSCAN Clustering on Curved Dataset")

# Highlight noise points (labeled as -1)
noise_points = X[labels == -1]
if len(noise_points) > 0:
    plt.scatter(noise_points[:, 0], noise_points[:, 1], c='red', marker='x', label='Noise')

plt.legend()
plt.show()
```

```python
# core point is a point that has at least min_samples points within eps distance
# border point is a point that has less than min_samples points within eps distance
# # noise point is a point that has less than min_samples points within eps distance
# 1. Cohesion (Intra-cluster Similarity)
# Definition:
# Cohesion measures how closely related the points within a single cluster are. A cluster with high cohesion means that all points in that cluster are close to each other.
```

```python
# 2. Separation (Inter-cluster Difference)
# Definition:
# Separation measures how distinct or well-separated different clusters are. A clustering solution with high separation means that clusters are well apart from each other.
# A good clustering algorithm maximizes separation,
# A good clustering algorithm balances both:
# ✔ Low cohesion (tight clusters) → Ensures meaningful groupings.
# ✔ High separation (well-separated clusters) → Ensures distinct clusters.

# If clusters have high cohesion but low separation, they may be too close to each other (overlapping).
# If clusters have low cohesion but high separation, they may be too sparse (less meaningful clusters).

```
