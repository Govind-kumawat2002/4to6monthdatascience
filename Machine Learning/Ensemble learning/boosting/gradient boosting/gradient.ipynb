```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import load_iris , load_breast_cancer, load_wine
import pandas as pd 
```

```python
x , y =load_wine(return_X_y=True)
```

```python
x 
```

```python
y 
```

```python
x =pd.DataFrame(x)
y =pd.DataFrame(y)
```

```python
x.head()
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2,random_state=21)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
gbc=GradientBoostingClassifier()
```

```python
gbc.fit(x_train,y_train)
```

```python
print("test data score",gbc.score(x_test,y_test))
print("train data score",gbc.score(x_train,y_train))
```

```python
gbc.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
print("confusion matrix",confusion_matrix(y_test,gbc.predict(x_test)))
```

```python
gbc.get_params()
```

```python
gbc=GradientBoostingClassifier(ccp_alpha=0.01
                               ,criterion='friedman_mse',
                               init=None,
                               learning_rate=0.1,
                               loss='log_loss',
                               max_depth=3,
                               max_features=None,
                               max_leaf_nodes=None,
                               min_impurity_decrease=0.01,
                               n_estimators=110,
                               random_state=None,
                               subsample=1.0,
                               tol=0.0001,
                               validation_fraction=0.1,
                               verbose=0,
                               warm_start=False)
```

```python
gbc.fit(x_train,y_train)
```

```python
print("test data score",gbc.score(x_test,y_test))
print("train data score",gbc.score(x_train,y_train))
```
