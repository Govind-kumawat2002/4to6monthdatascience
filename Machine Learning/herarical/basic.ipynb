```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
import pandas as pd
import scipy.cluster.hierarchy as shc
```

```python
path = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
```

```python
df =pd.read_csv(path)
```

```python
df.head()
```

```python
data =df.iloc[:,:-1].values
```

```python
data
```

```python
# dendrogram
# 
plt.figure(figsize=(10, 5)) 
linkage_data=shc.linkage(data,method='ward')
shc.dendrogram(linkage_data,labels=df.index)
plt.title("Agglomerative hierarchical clustering- dendrogram")
plt.xlabel("datapoint")
plt.ylabel("distance")
plt.show()
```

```python
num_cluster = 3
model =AgglomerativeClustering(n_clusters=num_cluster,metric='euclidean',linkage='ward') # >>>>>  min , moax , group , ward 
model.fit(data)
# model = AgglomerativeClustering(n_clusters=num_cluster, affinity='euclidean', linkage='ward')
```
