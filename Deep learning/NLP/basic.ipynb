## chat word treatement 

```python
# u = "you"
# b4 = "before"
# pls = "please"
# jk = "joke"
# ily = "ily"

chat_word_dict = {
    "u": "you",
    "r": "are",
    "b4": "before",
    "gr8": "great",
    "l8r": "later",
    "idk": "I don’t know",
    "btw": "by the way",
    "imo": "in my opinion",
    "ttyl": "talk to you later",
    "brb": "be right back",
    "wanna": "want to",
    "gonna": "going to",
    "lemme": "let me",
    "dunno": "don’t know",
    "thx": "thanks",
    "pls": "please",
    "ppl": "people",
    "smh": "shaking my head",
    "tbh": "to be honest",
    "omg": "oh my god",
    "lol": "laugh out loud",
    "rofl": "rolling on the floor laughing",
    "afk": "away from keyboard",
    "np": "no problem",
    "gtg": "got to go",
    "jk": "just kidding",
    "hbu": "how about you",
    "ily": "I love you",
    "ttys": "talk to you soon"
}
# chat_word_dict
```

```python
def normalize_chat_word(input):
    words =input.split()
    normalize_words=[chat_word_dict.get(word.lower(),word)for word in words]
    return " ".join(normalize_words)
    

```

```python
chat_text = "omg idk what u r talking about lol"
normalize_chat_word(input=chat_text)
```
