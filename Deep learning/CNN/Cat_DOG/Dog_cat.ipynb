```python
img_arr_lst = []
labels = []
for key , valu in path_dict.items(): ### cat or dog  keys cat or dog , valu ke value cat or dog images 
    for path in valu:
        # print(path)
        img_arr=cv2.imread(path) # imread aapki jo image usko read krta hai or array me convert krta hai
        resized_img_arr=cv2.resize(img_arr,(224,244))  # resize krta hai , 224 and 244 height and width size
        img_arr_lst.append(resized_img_arr) 
        labels.append(key)    # cat and dogs 


    
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(array_img, labels, test_size=0.2,random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
# 0 to 255
# 0 - 1
# -1  to +1 127.5
x_train=x_train/255  
x_test=x_test/255  # rgb 0 to 255 , b/w 0 to 1 ,||||||1 white or black

```

```python
from tensorflow.keras.models import Sequential  
```

```python
# from sklearn.ensemble import 
# model =RandomForestClassifier(





# )
```

```python
model =Sequential([
    # convolution layer
    layers.Conv2D(32,(3,3),activation='relu',input_shape=(244,224,3)), # convolution layer
    layers.MaxPooling2D((2,2)), # pooling layer , max pooling , 2 , 2 is kernel size
    layers.Conv2D(64,(3,3),activation='relu'), # convolution layer
    layers.MaxPooling2D((2,2)), # pooling layer , max pooling , 2 , 2 is kernel size
    layers.Conv2D(128,(3,3),activation='relu'), # convolution layer


    # Flattening layer 
    layers.Flatten(), # output hai vo 1 d 

    layers.Dense(64, activation='relu'),






    layers.Dense(1, activation='sigmoid') # simgoid 0 to 1
])

# backword propagation , optimizer , loss , metrics
model.compile(optimizer = 'adam',loss = 'binary_crossentropy',metrics = ['accuracy'])
```

```python
model.summary()
```

```python
history =model.fit(x_train, y_train, initial_epoch=2 , epochs=100 , batch_size=52,validation_data=(x_test,y_test))
```
